```python
#Split by Whitespace
import re
text = 'I\'m with you for the entire life in U.K.!'
words = re.split(r'\W+', text)
print(words)
```

##### When combined, \W+ matches one or more consecutive characters that are not letters, digits, or underscores. This can be useful for finding sequences of spaces, punctuation marks, or other special characters in a string.

```python
#Select Words
words = re.split(r'\W+', text)
print(words[:100])
```

```python
import string
import re
# split into words by white space
words = text.split()
print(words)
# prepare regex for char filtering
re_punc = re.compile('[%s]' % re.escape(string.punctuation))
print(re_punc)
# remove punctuation from each word
stripped = [re_punc.sub('', w) for w in words]
print(stripped[:100])
```

```python
# Normalizing Case

# split into words by white space
words = text.split()
# convert to lower case
words = [word.lower() for word in words]
print(words[:100])
```
